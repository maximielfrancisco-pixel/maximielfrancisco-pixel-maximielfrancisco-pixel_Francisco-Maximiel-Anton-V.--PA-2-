#### PROBLEM 1: REPRODUCIBLE NORMALIZATION PROBLEM

```python
# It will generate a reproducible random 5x5 array
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
# Get the mean of X
xbar = np.mean(X)
xbar
```

```python
# Get the standard deviation of X
stndrdev = np.std(X)
stndrdev
```

```python
# Normalize the array
X_normalized = (X - xbar)/stndrdev
X_normalized
```

```python
# Check the mean of the normalized array
np.mean(X_normalized)
```

```python
# Check the standard deviation of the normalized array
np.std(X_normalized)
```

```python
np.save("X_normalized",X_normalized)
```

#### PROBLEM 2: CUBES DIVISIBLE BY 4 PROBLEM

```python
# Create and cube the first 100 positive integers
C = (np.arange(1,101) ** 3).reshape(10,10)
C
```

```python
# Select values divisible by 4
div_by_4 = C[C % 4 == 0]
div_by_4
```

```python
C.shape
```

```python
div_by_4.size
```

```python
# Save the selected values
np.save("div_by_4.npy", div_by_4)
```

#### PROBLEM 3: ABOVE-MEAN SQUARES PROBLEM

```python
S = (np.arange(1,37) ** 2).reshape(6,6)
S
```

```python
S_mean = np.mean(S)
S_mean
```

```python
above_mean = S[S > S_mean]
above_mean
```

```python
above_mean.size
```

```python
np.save("above_mean", above_mean)
```
